# TF-IDF and Web Search Models Evaluation

This notebook explores the implementation of the `TfIdfBaseModel` and `TfIdfWebModel` modalities. 

### Goals
1. **Base Modality**: Compare the TF-IDF representation of the question directly against the answers using Cosine Similarity.
2. **Web Modality**: Query the web with the question to obtain context, and compare the TF-IDF representation of that search document against the answers.

```python
import warnings
warnings.filterwarnings('ignore')

from models.tfidf import TfIdfBaseModel, TfIdfWebModel

# Example Trivia Setup
question = "Which planet is known as the Red Planet?"
options = {
    '1': 'Venus',
    '2': 'Mars',
    '3': 'Jupiter',
    '4': 'Saturn'
}
```

### Evaluating the models

```python
base_model = TfIdfBaseModel()
base_answer = base_model.answer(question, options)
print(f"[Base TF-IDF Model] Selected Option ID: {base_answer} -> {options[str(base_answer)]}")

web_model = TfIdfWebModel()
web_doc = web_model.get_web_document(question)
print(f"\n[Web Document Retrieved]: \n\"{web_doc}\"\n")

web_answer = web_model.answer(question, options)
print(f"[Web TF-IDF Model] Selected Option ID: {web_answer} -> {options[str(web_answer)]}")
```
